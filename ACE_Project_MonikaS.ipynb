```python
SYSTEM_PROMPT = """
You are an AI Internship Assistant.

Your job is to:
1. Suggest internships based on user profile
2. Improve resume bullet points
3. Generate LinkedIn messages
4. Provide learning roadmap

Rules:
- Be specific and actionable
- No vague advice
- Output must be structured
- If unclear, ask questions

Always respond in this format:
Task:
Steps:
Final Answer:
"""
```

```python
def suggest_internships(user_input):
    return f"Suggested roles for: {user_input}"

def improve_resume(text):
    return f"Improved resume: {text}"

def linkedin_message(role):
    return f"Hi, I am interested in {role} internship..."

def roadmap(goal):
    return f"Step-by-step roadmap for {goal}"
```

```python
def run_agent(user_input):

    if "resume" in user_input:
        return improve_resume(user_input)

    elif "internship" in user_input:
        return suggest_internships(user_input)

    elif "linkedin" in user_input:
        return linkedin_message(user_input)

    elif "roadmap" in user_input:
        return roadmap(user_input)

    else:
        response = client.chat.completions.create(
            model="openai/gpt-4o-mini",
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_input}
            ]
        )
        return response.choices[0].message.content
```

```python
print(run_agent("I want internship in AI"))
print(run_agent("Improve my resume: built a project"))
print(run_agent("Write linkedin message for HR"))
print(run_agent("Give roadmap for full stack developer"))
```

```python
# V1 Prompt
v1 = "Help with internship"

# V2 Prompt
v2 = SYSTEM_PROMPT

print("V1:", v1)
print("V2:", v2)
```

```python
def suggest_internships(user_input):
    response = client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": f"Suggest internships for: {user_input}"}
        ]
    )
    return response.choices[0].message.content


def improve_resume(text):
    response = client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": """
You are a professional resume improvement assistant.

Rules:
1. Use only facts explicitly provided by the user.
2. Never invent technologies, tools, responsibilities,
   achievements, metrics, percentages, or user counts.
3. Improve grammar, clarity, and professional wording.
4. If details are missing, write a truthful bullet using
   the available information.
5. Suggest questions that could help quantify the achievement,
   but never invent answers.
6. Separate the improved bullet from suggestions for more details.
"""
            },
            {
                "role": "user",
                "content": f"Improve this resume content using the rules above:\n{text}"
            }
        ],
        temperature=0.2
    )
    return response.choices[0].message.content


def linkedin_message(role):
    response = client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": f"Write LinkedIn message for: {role}"}
        ]
    )
    return response.choices[0].message.content


def roadmap(goal):
    response = client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": f"Create roadmap for: {goal}"}
        ]
    )
    return response.choices[0].message.content
```

```python

print(run_agent("Improve my resume: built a chatbot"))
```

```python
def run_agent(user_input):
    text = user_input.lower()

    if "linkedin" in text or "recruiter" in text or "hr message" in text:
        print("Tool Used: LinkedIn Generator")
        return linkedin_message(user_input)

    elif "resume" in text or "cv" in text:
        print("Tool Used: Resume Improver")
        return improve_resume(user_input)

    elif "roadmap" in text or "learning path" in text:
        print("Tool Used: Roadmap Planner")
        return roadmap(user_input)

    elif "internship" in text or "intern" in text:
        print("Tool Used: Internship Suggester")
        return suggest_internships(user_input)

    else:
        print("Tool Used: General AI")
        response = client.chat.completions.create(
            model="openai/gpt-4o-mini",
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_input}
            ]
        )
        return response.choices[0].message.content
```

```python

print(run_agent("Write a LinkedIn message for a software internship"))
```

```python
print(run_agent("I am a CSE student, suggest internships"))
print(run_agent("Improve my resume: built a chatbot"))
print(run_agent("Write linkedin message for software internship"))
print(run_agent("Give roadmap for AI engineer"))
```
